```python
import requests
from bs4 import BeautifulSoup
import time
import csv

def fetch_search_page(url, headers):
    resp = requests.get(url, headers=headers)
    if resp.status_code != 200:
        print("Failed to fetch", url, resp.status_code)
        return None
    return resp.text

def parse_review_snippets(html):
    soup = BeautifulSoup(html, "html.parser")
    # Example: find elements which look like review snippets
    snippets = []
    for span in soup.select("div.g"):  # generic Google result block
        text = span.get_text().strip()
        if "review" in text.lower(): 
            snippets.append(text)
    return snippets

def main():
    url = "https://www.google.com/search?q=Ek+Deewane+Ki+Deewaniyat+movie+reviews"
    headers = {"User-Agent":"Mozilla/5.0 (Windows NT 10.0; Win64; x64)"}
    html = fetch_search_page(url, headers)
    if html:
        reviews = parse_review_snippets(html)
        print("Found snippets:", len(reviews))
        with open("movie_review_snippets.csv", "w", newline="", encoding="utf-8") as f:
            writer = csv.writer(f)
            writer.writerow(["snippet"])
            for s in reviews:
                writer.writerow([s])
    else:
        print("No html retrieved.")

if __name__ == "__main__":
    main()
```
