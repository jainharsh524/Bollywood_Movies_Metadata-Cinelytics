```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
import time
import numpy as np

# Step 1: Load your base dataset
df = pd.read_csv('bollywood_2015_2025.csv')

# Ensure there's a 'Year' column
if 'Year' not in df.columns:
    raise ValueError("CSV must have a 'Year' column.")

# Step 2: Prepare list of movies (Title + Year)
movie_titles = df[['Title', 'Year']].dropna().values.tolist()

base_url = "https://www.boxofficeindia.com"
scraped_data = []

# Step 3: Loop through movies
for title, year in movie_titles:
    try:
        print(f"🔍 Searching for: {title} ({int(year)})")

        # BoxOfficeIndia search URL
        search_url = f"{base_url}/search.php?txtSearchStr={title.replace(' ', '+')}&search_type=movies"
        response = requests.get(search_url, headers={"User-Agent": "Mozilla/5.0"})
        soup = BeautifulSoup(response.text, 'html.parser')

        # Step 3A: Find all search result links
        movie_links = soup.find_all('a', href=True)
        movie_page_link, movie_year = None, None

        # Exact match with year
        for link in movie_links:
            link_text = link.text.strip()
            if link_text.lower().startswith(title.strip().lower()):
                if f"({int(year)})" in link_text or str(int(year)) in link_text:
                    movie_page_link = link['href']
                    # Extract year from text if available
                    if '(' in link_text and ')' in link_text:
                        movie_year = link_text.split('(')[-1].split(')')[0]
                    break

        if not movie_page_link:
            print(f"❌ No exact match found for {title} ({int(year)})")
            scraped_data.append({
                'Title': title,
                'Year': year,
                'Budget': np.nan,
                'India Gross': np.nan,
                'Worldwide Gross': np.nan
            })
            continue

        # Step 3B: Build movie URL
        if not movie_page_link.startswith('http'):
            movie_page_link = base_url + '/' + movie_page_link.lstrip('/')

        # Step 4: Request movie page
        movie_page = requests.get(movie_page_link, headers={"User-Agent": "Mozilla/5.0"})
        movie_soup = BeautifulSoup(movie_page.text, 'html.parser')

        # Step 5: Robust parsing of Budget, India Gross, Worldwide
        budget = india_gross = worldwide_gross = None
        details_table = movie_soup.find('table', class_='mviedtailstbe')

        if details_table:
            for row in details_table.find_all('tr'):
                tds = row.find_all('td')
                if len(tds) >= 3:
                    key = tds[0].get_text(" ", strip=True).lower()
                    # Extract text from 3rd td ignoring rupee icon or spans
                    val = ''.join([s for s in tds[2].stripped_strings if s not in ['₹', '', '\xa0']])
                    if 'budget' in key:
                        budget = val
                    elif 'india gross' in key:
                        india_gross = val
                    elif 'worldwide' in key:
                        worldwide_gross = val

        scraped_data.append({
            'Title': title,
            'Year': movie_year or year,
            'Budget': budget,
            'India Gross': india_gross,
            'Worldwide Gross': worldwide_gross
        })

        print(f"✅ {title} ({movie_year or year}) | Budget: {budget} | India Gross: {india_gross} | Worldwide: {worldwide_gross}")
        time.sleep(1.5)  # polite delay

    except Exception as e:
        print(f"⚠ Error scraping {title}: {e}")
        scraped_data.append({
            'Title': title,
            'Year': year,
            'Budget': np.nan,
            'India Gross': np.nan,
            'Worldwide Gross': np.nan
        })

# Step 6: Convert scraped data to DataFrame
df_scraped = pd.DataFrame(scraped_data)

# Step 7: Merge scraped data with original CSV
df_final = df.merge(df_scraped, on=['Title', 'Year'], how='left', suffixes=('', '_scraped'))

# Step 8: Save updated CSV
df_final.to_csv('bollywood_2015_2025_updated.csv', index=False)

print("\n🎉 Scraping complete! Data saved as 'bollywood_2015_2025_updated.csv'")
```

```python
import time
import pandas as pd
import numpy as np
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.chrome.service import Service
from webdriver_manager.chrome import ChromeDriverManager

# === Chrome setup with webdriver-manager (auto-matched version) ===
chrome_options = Options()
chrome_options.add_argument("--headless")  # remove to see browser
chrome_options.add_argument("--no-sandbox")
chrome_options.add_argument("--disable-dev-shm-usage")

service = Service(ChromeDriverManager().install())
driver = webdriver.Chrome(service=service, options=chrome_options)

# === Load CSV ===
df = pd.read_csv("bollywood_2015_2025.csv")
movie_titles = df["Title"].dropna().unique().tolist()
base_url = "https://www.boxofficeindia.com"

scraped_data = []

for title in movie_titles:
    try:
        print(f"🔍 Searching for: {title}")
        # Search page
        search_url = f"{base_url}/search.php?txtSearchStr={title.replace(' ', '+')}&search_type=movies"
        driver.get(search_url)
        time.sleep(2)

        # Get all links from search results
        links = driver.find_elements(By.TAG_NAME, "a")
        movie_page_link, movie_year = None, None

        for link in links:
            text = link.text.strip()
            if text.lower().startswith(title.strip().lower()):
                movie_page_link = link.get_attribute("href")
                # Extract year if available like "Alone (2015)"
                if "(" in text and ")" in text:
                    movie_year = text.split("(")[-1].split(")")[0]
                break

        if not movie_page_link:
            print(f"❌ No exact match for {title}")
            scraped_data.append({
                "Title": title,
                "Year": np.nan,
                "Budget": np.nan,
                "India Gross": np.nan,
                "Worldwide Gross": np.nan
            })
            continue

        # === Open movie page ===
        driver.get(movie_page_link)
        time.sleep(3)

        # Extract financials from table
        rows = driver.find_elements(By.CSS_SELECTOR, "table.mviedtailstbe tr")
        budget, india_gross, worldwide_gross = None, None, None

        for row in rows:
            cols = row.find_elements(By.TAG_NAME, "td")
            if len(cols) >= 3:
                key = cols[0].text.strip().lower()
                val = cols[2].text.strip()
                if "budget" in key:
                    budget = val
                elif "india gross" in key:
                    india_gross = val
                elif "worldwide gross" in key:
                    worldwide_gross = val

        scraped_data.append({
            "Title": title,
            "Year": movie_year,
            "Budget": budget,
            "India Gross": india_gross,
            "Worldwide Gross": worldwide_gross
        })

        print(f"✅ {title} ({movie_year}) | Budget: {budget} | India Gross: {india_gross} | Worldwide: {worldwide_gross}")
        time.sleep(1)

    except Exception as e:
        print(f"⚠ Error scraping {title}: {e}")
        scraped_data.append({
            "Title": title,
            "Year": np.nan,
            "Budget": np.nan,
            "India Gross": np.nan,
            "Worldwide Gross": np.nan
        })

# === Save Results ===
df_scraped = pd.DataFrame(scraped_data)
df_final = df.merge(df_scraped, on="Title", how="left", suffixes=("", "_scraped"))
df_final.to_csv("bollywood_2015_2025_selenium.csv", index=False)

print("\n🎉 Scraping complete! Saved as 'bollywood_2015_2025_selenium.csv'")
driver.quit()
```

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
import numpy as np
import re
from difflib import SequenceMatcher

# -------- Config --------
file_path = "Bollywood - bollywood_with_genres_gpt_partial - bollywood_with_genres_gpt_partial (1).csv"
output_path = "bollywood_with_sacnilk_filled.csv"
url = "https://www.sacnilk.com/entertainmenttopbar2/RecentMoviesCollection"
headers = {"User-Agent": "Mozilla/5.0"}
# ------------------------

def parse_number_to_float(s):
    if s is None:
        return np.nan
    s = str(s).strip()
    if s in ["", "--", "None", "nan"]:
        return np.nan
    cleaned = re.sub(r"[^\d.\-]", "", s)
    if cleaned in ["", ".", "-"]:
        return np.nan
    try:
        return float(cleaned)
    except:
        return np.nan

def crore_to_rupees(value_cr):
    if pd.isna(value_cr):
        return np.nan
    try:
        return float(value_cr) * 10000000  # 1 crore = 10,000,000
    except:
        return np.nan

# Load dataset
df = pd.read_csv(file_path)

# Fetch and parse page
resp = requests.get(url, headers=headers)
resp.raise_for_status()
soup = BeautifulSoup(resp.text, "html.parser")
table = soup.find("table")
if table is None:
    raise RuntimeError("Table not found on Sacnilk page")

# Correct columns:
# index 2 = Worldwide (Cr)
# index 4 = India Gross (Cr)
movie_data = {}
for row in table.find_all("tr")[1:]:
    tds = row.find_all("td")
    cols = [td.text.strip().replace("\xa0", " ") for td in tds]
    if len(cols) < 5:
        continue

    title = cols[1].lower().strip()
    worldwide_cr = parse_number_to_float(cols[2])
    india_gross_cr = parse_number_to_float(cols[4])

    movie_data[title] = {
        "Worldwide (₹)": crore_to_rupees(worldwide_cr),
        "India Gross (₹)": crore_to_rupees(india_gross_cr),
    }

print("\n📊 Scraped sample:")
for i, (k, v) in enumerate(movie_data.items()):
    print(k, "→", v)
    if i >= 4:
        break

# Smart title matching
def find_best_match(title):
    if not isinstance(title, str):
        return None
    title_clean = title.lower().strip()
    best_match = None
    best_score = 0.0
    for candidate in movie_data.keys():
        score = SequenceMatcher(None, title_clean, candidate).ratio()
        if score > best_score:
            best_score = score
            best_match = candidate
    # accept only if similarity ≥ 0.9 (very close)
    if best_score >= 0.9:
        return movie_data[best_match]
    else:
        return None

# Fill missing safely
def is_empty(x):
    if pd.isna(x):
        return True
    s = str(x).strip()
    return s == "" or s == "--" or s.lower() == "nan"

updated_movies = []
india_updates = []
world_updates = []

for idx, row in df.iterrows():
    title = row.get("Title", "")
    india_value = row.get("India Gross", None)
    world_value = row.get("Worldwide Gross_scraped", None)

    data = find_best_match(title)
    india_new = india_value
    world_new = world_value
    updated = False

    if data:
        scraped_india = data.get("India Gross (₹)")
        scraped_world = data.get("Worldwide (₹)")

        if is_empty(india_value) and not pd.isna(scraped_india) and scraped_india > 0:
            india_new = scraped_india
            updated = True

        if is_empty(world_value) and not pd.isna(scraped_world) and scraped_world > 0:
            world_new = scraped_world
            updated = True

    if updated:
        updated_movies.append(title)

    india_updates.append(india_new)
    world_updates.append(world_new)

df["India Gross"] = india_updates
df["Worldwide Gross_scraped"] = world_updates
df.to_csv(output_path, index=False)

print(f"\n✅ Saved updated file → {output_path}")
if updated_movies:
    print(f"\n🎬 {len(updated_movies)} movies updated:")
    for m in updated_movies:
        print(" -", m)
else:
    print("\nℹ️ No movies updated (no close matches found).")
```

```python
import pandas as pd
from difflib import SequenceMatcher

# === Step 1: Load datasets ===
movies_df = pd.read_csv("bollywood_movies.csv")
dataset_df = pd.read_csv("Dataset - bollywood_with_sacnilk_filled.csv")

# === Step 2: Clean column names ===
movies_df.columns = movies_df.columns.str.strip().str.lower()
dataset_df.columns = dataset_df.columns.str.strip()

# === Step 3: Add column if missing ===
if 'Review_positive' not in dataset_df.columns:
    dataset_df['Review_positive'] = None

# === Step 4: Define helper function ===
def similarity(a, b):
    """Compute similarity ratio between two strings."""
    return SequenceMatcher(None, str(a).lower(), str(b).lower()).ratio()

# === Step 5: Fill missing Review_positive values only ===
for i, row in dataset_df.iterrows():
    if pd.isna(row['Review_positive']) or row['Review_positive'] == '':
        title = row['Title']
        best_match = None
        best_ratio = 0

        for _, m_row in movies_df.iterrows():
            ratio = similarity(title, m_row['title'])
            if ratio > best_ratio:
                best_match = m_row
                best_ratio = ratio

        if best_ratio >= 0.9:
            dataset_df.at[i, 'Review_positive'] = best_match['rating']/10
            print(f"✅ Filled for '{title}' (match: '{best_match['title']}' – {best_ratio:.2f})")

# === Step 6: Save updated dataset ===
dataset_df.to_csv("Dataset_bollywood_with_review_positive.csv", index=False)
print("\n🎬 Updated file saved as 'Dataset_bollywood_with_review_positive.csv'")
```

```python
import pandas as pd
import numpy as np
import random

# === Step 1: Load dataset ===
df = pd.read_csv("Dataset_bollywood_with_review_sentiments.csv")

# === Step 2: Define function for controlled random generation ===
def generate_balanced_random():
    """
    Generate random values between 0.2 and 0.8.
    Keep 0.4–0.6 more common.
    Make 0.2–0.4 and 0.7–0.8 less frequent.
    """
    r = random.random()
    if r < 0.15:  # 15% chance: lower range
        return round(random.uniform(0.2, 0.4), 2)
    elif r < 0.85:  # 70% chance: mid range
        return round(random.uniform(0.4, 0.6), 2)
    else:  # 15% chance: upper range
        return round(random.uniform(0.6, 0.8), 2)

# === Step 3: Fill missing values ===
if 'Trailer_comment_positive' not in df.columns:
    raise ValueError("Column 'Trailer_comment_positive' not found!")

for i, val in df['Trailer_comment_positive'].items():
    if pd.isna(val) or str(val).strip() == '':
        df.at[i, 'Trailer_comment_positive'] = generate_balanced_random()

# === Step 4: Save updated dataset ===
df.to_csv("Dataset_bollywood_with_trailer_comments_filled.csv", index=False)

print("✅ Missing 'Youtube Trailer_comment_positive' values filled successfully.")
print("📁 Saved as: Dataset_bollywood_with_trailer_comments_filled.csv")
```

```python
import pandas as pd
import re

# --- Load dataset ---
file_path = "Dataset_bollywood_with_trailer_comments_filled.csv"  # adjust if needed
df = pd.read_csv(file_path)

# --- Detect India Gross / Collection column ---
gross_columns = [col for col in df.columns if 'gross' in col.lower() or 'collection' in col.lower() or 'box' in col.lower()]
gross_col = gross_columns[0] if gross_columns else None

# --- Conversion Function ---
def convert_indian_number(val):
    """
    Convert Indian number strings (like '20,00,00,000') or plain numeric values
    into float rupees.
    """
    if pd.isna(val):
        return 0.0

    val = str(val).strip()

    # Remove any ₹ symbol, commas, spaces
    clean_val = re.sub(r'[₹, ]', '', val)

    # If it’s just a number, return as float
    if clean_val.replace('.', '', 1).isdigit():
        return float(clean_val)

    # If pattern like 20,00,00,000 (already handled by removing commas)
    try:
        return float(clean_val)
    except:
        return 0.0

# --- Apply Conversion ---
df[gross_col + "_numeric"] = df[gross_col].apply(convert_indian_number)

# --- Check comparison (top 10 highest collections) ---
top_movies = df[[gross_col, gross_col + "_numeric"]].sort_values(by=gross_col + "_numeric", ascending=False).head(10)
print(top_movies)
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import re

# --- Load dataset ---
file_path = "Dataset_bollywood_with_trailer_comments_filled.csv"   # update if needed
df = pd.read_csv(file_path)

# --- Style settings ---
plt.style.use('dark_background')
fg_color = '#FFD700'
line_color = '#39FF14'   # neon green


# --- Identify Year and Worldwide columns ---
year_columns = [col for col in df.columns if 'year' in col.lower()]
worldwide_columns = [col for col in df.columns if 'worldwide gross_scraped' in col.lower()]

year_col = year_columns[0]
world_col = worldwide_columns[0]


# --- Clean Year Column ---
df[year_col] = df[year_col].astype(str).str.extract(r'(\d{4})').astype(float)


# --- Function: Convert values like "20,00,00,000" to numeric ---
def convert_indian_number(val):
    if pd.isna(val):
        return 0.0
    val = str(val).strip()

    # remove indian commas and symbols
    clean_val = re.sub(r'[₹, ]', '', val)

    # convert to float if possible
    try:
        return float(clean_val)
    except:
        return 0.0


# --- Apply conversion ---
df[world_col + "_numeric"] = df[world_col].apply(convert_indian_number)


# --- Group by year and sum worldwide gross ---
yearly_worldwide = df.groupby(year_col)[world_col + "_numeric"].sum().reset_index()


# --- Plot neon line graph ---
plt.figure(figsize=(10, 6))
plt.plot(
    yearly_worldwide[year_col], 
    yearly_worldwide[world_col + "_numeric"],
    color=line_color, linewidth=2, marker='o'
)

plt.title("🌍 Total Worldwide Gross by Year", color=fg_color, fontsize=16, pad=15)
plt.xlabel("Year", color=fg_color, fontsize=12)
plt.ylabel("Worldwide Gross (Sum)", color=fg_color, fontsize=12)
plt.xticks(rotation=45, color=fg_color)
plt.yticks(color=fg_color)

plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import plotly.graph_objects as go
import re

# Load dataset
file_path = "Dataset_bollywood_with_trailer_comments_filled.csv"  # change path if needed
df = pd.read_csv(file_path)

# Identify year & worldwide columns
year_columns = [col for col in df.columns if 'year' in col.lower() or 'release' in col.lower()]
worldwide_columns = [col for col in df.columns if 'world' in col.lower() or 'total gross' in col.lower()]

year_col = year_columns[0]
world_col = worldwide_columns[0]

# Clean year
df[year_col] = df[year_col].astype(str).str.extract(r'(\d{4})').astype(float)

# Convert Indian comma numbers like "129,00,00,000"
def convert_indian_number(val):
    if pd.isna(val):
        return 0.0
    val = str(val).strip()
    clean_val = re.sub(r'[₹, ]', '', val)
    try:
        return float(clean_val)
    except:
        return 0.0

df[world_col + "_numeric"] = df[world_col].apply(convert_indian_number)

# Group by year
yearly_worldwide = df.groupby(year_col)[world_col + "_numeric"].sum().reset_index()

# Create interactive plot
fig = go.Figure()

# BAR graph 
fig.add_trace(go.Bar(
    x=yearly_worldwide[year_col],
    y=yearly_worldwide[world_col + "_numeric"],
    name="Worldwide Gross (Bars)",
    marker=dict(color='cyan'),
    opacity=0.6
))

# LINE graph
fig.add_trace(go.Scatter(
    x=yearly_worldwide[year_col],
    y=yearly_worldwide[world_col + "_numeric"],
    name="Worldwide Gross (Line)",
    mode='lines+markers',
    line=dict(color='lime', width=3),
    marker=dict(size=8)
))

# Layout
fig.update_layout(
    title="🌍 Worldwide Gross by Year (Interactive Line + Bar Combo)",
    xaxis_title="Year",
    yaxis_title="Worldwide Gross (Total)",
    template="plotly_dark",
    title_font=dict(size=22, color="gold"),
    font=dict(color="gold"),
    legend=dict(font=dict(color="gold"))
)

fig.show()
```

```python
import pandas as pd
import re

# Load dataset
file_path = "Dataset_bollywood_with_trailer_comments_filled.csv"
df = pd.read_csv(file_path)

# Identify budget and India gross columns
budget_cols = [c for c in df.columns if 'budget' in c.lower()]
india_cols = [c for c in df.columns if ('india' in c.lower() and 'gross' in c.lower())]

# Fallback if exact India gross is not found
if not india_cols:
    india_cols = [c for c in df.columns if 'gross' in c.lower() or 'collection' in c.lower()]

budget_col = budget_cols[0]
india_col = india_cols[0]

# Function: convert Indian comma numbers to numeric
def convert_indian_number(val):
    if pd.isna(val):
        return 0.0
    val = str(val).strip()
    clean = re.sub(r"[₹, ]", "", val)
    try:
        return float(clean)
    except:
        return 0.0

# Convert columns to numeric
df["budget_num"] = df[budget_col].apply(convert_indian_number)
df["india_gross_num"] = df[india_col].apply(convert_indian_number)

# Verdict conditions
def get_verdict(row):
    gross = row["india_gross_num"]
    budget = row["budget_num"]

    # NEW CONDITION → If budget is missing or zero
    if budget == 0:
        return ""

    # 1. If India Gross < 3 crore (3 crore = 3 × 1e7)
    if gross < 3 * 1e7:
        return "Disaster"

    # 2. If India Gross < 118% of budget
    if gross < 1.18 * budget:
        return "Flop"

    # 3. If between 118% and 136%
    if 1.18 * budget <= gross < 1.36 * budget:
        return "Average"

    # 4. If between 136% and 200%
    if 1.36 * budget <= gross < 2.0 * budget:
        return "Hit"

    # 5. If above 200%
    if gross >= 2.0 * budget:
        return "Superhit / Blockbuster"

    return ""

# Apply verdict function
df["Verdict"] = df.apply(get_verdict, axis=1)

# Optional: Save output
df.to_csv("Dataset_with_Verdict.csv", index=False)

df[["budget_num", "india_gross_num", "Verdict"]].head()
```
